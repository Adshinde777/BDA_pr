```python
#step1 : import libraries
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
#step2 : create dataset
x=[
      [1, 2], [2, 1], [2, 3],
      [8, 8], [9, 8], [8, 9],
      [15, 2], [16, 3], [15, 4]
]
```

```python
#step3 : create k-maens model
model = KMeans(
    n_clusters= 3,
    random_state= 42,
    n_init= 10
)
```

```python
#step4 : apply k-means clustering
model.fit(x)
```

```python
#step5 : Display cluster labels
print("Cluster labels : ")
print(model.labels_)
```

```python
#step6 : display centroids
print("\n Cluster centroids : ")
print(model.cluster_centers_)
```

```python
#step7 : visualize clusters
plt.scatter(
    [p[0] for p in x],
    [p[1] for p in x],
    c= model.labels_

)

plt.scatter(
    model.cluster_centers_[:, 0],
    model.cluster_centers_[:, 1],
    marker = "x",
    s= 200,
    label= "Centroids"

)

plt.title("K-Means Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.show()
```
